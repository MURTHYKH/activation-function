```python
import numpy as np
import pandas as pd
```

```python
x = np.arange(-10, 11)
print(x)
```

##TASK-1

#BINARY STEP

```python
def binary_step(x):
    return np.where(x >= 0, 1, 0)

binary_output = binary_step(x)

print(binary_output)
```

#SIGMOID FUNCTION

```python
def sigmoid(x):
    return 1/(1 + np.exp(-x))

sigmoid_output = sigmoid(x)

print(sigmoid_output)
```

#TANH FUNCTION

```python
def tanh(x):
    return np.tanh(x)

tanh_output = tanh(x)

print(tanh_output)
```

#ReLU FUNCTION

```python
def relu(x):
    return np.maximum(0, x)

relu_output = relu(x)

print(relu_output)
```

#LEAKY ReLU

```python
def leaky_relu(x):
    return np.where(x > 0, x, 0.01*x)

leaky_output = leaky_relu(x)

print(leaky_output)
```

#ELU

```python
def elu(x):
    return np.where(x > 0, x, np.exp(x)-1)

elu_output = elu(x)

print(elu_output)
```

#SoftPlus

```python
def softplus(x):
    return np.log(1 + np.exp(x))

softplus_output = softplus(x)

print(softplus_output)
```

#SWISH FUNCTION

```python
def swish(x):
    return x/(1 + np.exp(-x))

swish_output = swish(x)

print(swish_output)
```

```python
table = pd.DataFrame({
    "Input": x,
    "Binary Step": binary_output,
    "Sigmoid": sigmoid_output,
    "Tanh": tanh_output,
    "ReLU": relu_output,
    "Leaky ReLU": leaky_output,
    "ELU": elu_output,
    "Softplus": softplus_output,
    "Swish": swish_output
})

print(table)
```

#TASK-3

```python
x = np.linspace(-10, 10, 500)
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s*(1-s)

sigmoid_output = sigmoid_derivative(x)
```

```python
def tanh_derivative(x):
    return 1 - np.tanh(x)**2

tanh_output = tanh_derivative(x)
```

```python
def relu_derivative(x):
    return np.where(x > 0, 1, 0)

relu_output = relu_derivative(x)
```

```python
def leaky_relu_derivative(x):
    return np.where(x > 0, 1, 0.01)

leaky_output = leaky_relu_derivative(x)
```

```python
plt.figure(figsize=(10,6))

plt.plot(x, sigmoid_output, label='Sigmoid Derivative')
plt.plot(x, tanh_output, label='Tanh Derivative')
plt.plot(x, relu_output, label='ReLU Derivative')
plt.plot(x, leaky_output, label='Leaky ReLU Derivative')

plt.title("Derivative of Activation Functions")
plt.xlabel("Input")
plt.ylabel("Derivative")

plt.grid(True)
plt.legend()

plt.show()
```

```python
table = pd.DataFrame({
    "Activation Function": [
        "Sigmoid",
        "Tanh",
        "ReLU",
        "Leaky ReLU"
    ],
    "Derivative Formula": [
        "σ(x)(1-σ(x))",
        "1-tanh²(x)",
        "1 if x>0 else 0",
        "1 if x>0 else 0.01"
    ]
})

print(table)
```

Sigmoid: The derivative is highest near the center and approaches zero for very large positive or negative inputs, which can lead to the vanishing gradient problem.

Tanh: Similar to sigmoid, but it is zero-centered. Its derivative also becomes very small for large input values.

ReLU: The derivative is 1 for positive inputs and 0 for negative inputs. This enables fast learning for positive values but may cause dead neurons for negative inputs.

Leaky ReLU: Similar to ReLU, but instead of having a derivative of 0 for negative inputs, it has a small value (0.01). This helps prevent the dead neuron problem.

```python
x = np.random.uniform(-20, 20, 1000)
```

```python
def sigmoid(x):
    return 1/(1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)
```

```python
sigmoid_output = sigmoid(x)
tanh_output = tanh(x)
relu_output = relu(x)
```

```python
plt.figure(figsize=(10,6))

plt.scatter(x, sigmoid_output, s=10, label="Sigmoid")
plt.scatter(x, tanh_output, s=10, label="Tanh")
plt.scatter(x, relu_output, s=10, label="ReLU")

plt.title("Vanishing Gradient Analysis")
plt.xlabel("Input")
plt.ylabel("Output")

plt.grid(True)
plt.legend()

plt.show()
```

Sigmoid and Tanh suffer from saturation and vanishing gradients for large input values, making training slower in deep networks. ReLU avoids the vanishing gradient problem for positive inputs but can produce dead neurons for negative inputs. These characteristics influence the choice of activation function in deep learning models.

#TASK-6

```python
import time
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import Dense, Flatten, LeakyReLU
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
x_train = x_train / 255.0
x_test = x_test / 255.0
```

#MODEL-A (SIGMOID)

```python
model_A = Sequential()
model_A.add(Flatten(input_shape=(28,28)))
model_A.add(Dense(units=128,activation='sigmoid'))
model_A.add(Dense(units=10,activation='softmax'))
```

```python
model_A.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
start_time = time.time()
history_sigmoid = model_A.fit(x_train,y_train,epochs=10,batch_size=32,validation_split=0.2)
end_time = time.time()

time_sigmoid = end_time - start_time
```

```python
loss_sigmoid, acc_sigmoid = model_A.evaluate(x_test, y_test)

val_acc_sigmoid = max(history_sigmoid.history['val_accuracy'])

print("Accuracy :", acc_sigmoid)
print("Loss :", loss_sigmoid)
print("Validation Accuracy :", val_acc_sigmoid)
print("Training Time :", time_sigmoid)
```

#MODEL-B(TANH)

```python
model_B = Sequential()
model_B.add(Flatten(input_shape=(28,28)))
model_B.add(Dense(units=128,activation='tanh'))
model_B.add(Dense(units=10,activation='softmax'))
```

```python
model_B.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])

start_time = time.time()
history_tanh = model_B.fit(x_train,y_train,epochs=10,batch_size=32,validation_split=0.2)
end_time = time.time()

time_tanh = end_time - start_time
```

```python
loss_tanh, acc_tanh = model_B.evaluate(x_test, y_test)

val_acc_tanh = max(history_tanh.history['val_accuracy'])

print("Accuracy :", acc_tanh)
print("Loss :", loss_tanh)
print("Validation Accuracy :", val_acc_tanh)
print("Training Time :", time_tanh)
```

#MODEL-C(ReLU)

```python
model_C = Sequential()
model_C.add(Flatten(input_shape=(28,28)))
model_C.add(Dense(units=128,activation='relu'))
model_C.add(Dense(units=10,activation='softmax'))
```

```python
model_C.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])

start_time = time.time()
history_relu = model_C.fit(x_train,y_train,epochs=10,batch_size=32,validation_split=0.2)
end_time = time.time()

time_relu = end_time - start_time
```

#TASK-10

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time

from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import Dense, Flatten, LeakyReLU

from sklearn.metrics import (
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    accuracy_score
)
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
model_relu = Sequential()

model_relu.add(Flatten(input_shape=(28,28)))

model_relu.add(Dense(
    units=128,
    activation='relu'
))

model_relu.add(Dense(
    units=10,
    activation='softmax'
))
```

```python
model_relu.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
start_time = time.time()

history_relu = model_relu.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)

end_time = time.time()

time_relu = end_time - start_time
```

```python
loss_relu, test_acc_relu = model_relu.evaluate(x_test, y_test)

train_acc_relu = history_relu.history['accuracy'][-1]

val_acc_relu = history_relu.history['val_accuracy'][-1]
```

```python
y_pred = model_relu.predict(x_test)

y_pred = np.argmax(y_pred, axis=1)
```

```python
precision_relu = precision_score(
    y_test,
    y_pred,
    average='weighted'
)

recall_relu = recall_score(
    y_test,
    y_pred,
    average='weighted'
)

f1_relu = f1_score(
    y_test,
    y_pred,
    average='weighted'
)

cm = confusion_matrix(y_test, y_pred)

print("Training Accuracy :", train_acc_relu)
print("Validation Accuracy :", val_acc_relu)
print("Test Accuracy :", test_acc_relu)
print("Precision :", precision_relu)
print("Recall :", recall_relu)
print("F1 Score :", f1_relu)
print("Training Time :", time_relu)

print("\nConfusion Matrix\n")
print(cm)
```

```python
plt.figure(figsize=(8,6))

plt.imshow(cm)

plt.title("Confusion Matrix - ReLU")

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.colorbar()

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(history_relu.history['loss'], label='Training Loss')
plt.plot(history_relu.history['val_loss'], label='Validation Loss')

plt.title("Loss Curve - ReLU")

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.legend()

plt.grid(True)

plt.show()
```
